```python
def calculate_epf_balance(
    epfBalance,
    avgDividend,
    age,
    retirementAge,
    salary,
    bonus,
    increment,
    eeContribution,
    erContribution,
    monthlyWithdrawal,
    annualWithdrawal
):
    # Convert percentages
    avgDividend /= 100
    increment /= 100
    eeContribution /= 100
    erContribution /= 100

    while age < retirementAge:

        # Annual contributions
        employeeContribution = salary * 12 * eeContribution
        employerContribution = salary * 12 * erContribution

        totalContribution = employeeContribution + employerContribution

        # Bonus
        bonusContribution = bonus * (eeContribution + erContribution)

        # Withdrawals
        totalWithdrawal = (monthlyWithdrawal * 12) + annualWithdrawal

        # Add contributions and withdrawals
        epfBalance += totalContribution
        epfBalance += bonusContribution
        epfBalance -= totalWithdrawal

        # Dividend
        dividend = epfBalance * avgDividend
        epfBalance += dividend

        # Salary increment
        salary *= (1 + increment)

        # Next year
        age += 1

    return epfBalance
```

```python
import ipywidgets as widgets
from IPython.display import display, HTML
```

```python
def display_epf_calculator():
    # Create input widgets
    epfBalance_w = widgets.FloatText(value=100000.0, description='EPF Balance (RM):', style={'description_width': 'initial'})
    avgDividend_w = widgets.FloatSlider(value=6.0, min=0.0, max=10.0, step=0.1, description='Avg. Dividend (%):', style={'description_width': 'initial'})
    age_w = widgets.IntSlider(value=30, min=18, max=60, step=1, description='Current Age:', style={'description_width': 'initial'})
    retirementAge_w = widgets.IntSlider(value=55, min=40, max=70, step=1, description='Retirement Age:', style={'description_width': 'initial'})
    salary_w = widgets.FloatText(value=5000.0, description='Monthly Salary (RM):', style={'description_width': 'initial'})
    bonus_w = widgets.FloatText(value=0.0, description='Annual Bonus (RM):', style={'description_width': 'initial'})
    increment_w = widgets.FloatSlider(value=3.0, min=0.0, max=10.0, step=0.1, description='Annual Increment (%):', style={'description_width': 'initial'})
    eeContribution_w = widgets.FloatSlider(value=11.0, min=0.0, max=30.0, step=0.1, description='Employee Contrib. (%):', style={'description_width': 'initial'})
    erContribution_w = widgets.FloatSlider(value=12.0, min=0.0, max=30.0, step=0.1, description='Employer Contrib. (%):', style={'description_width': 'initial'})
    monthlyWithdrawal_w = widgets.FloatText(value=0.0, description='Monthly Withdrawal (RM):', style={'description_width': 'initial'})
    annualWithdrawal_w = widgets.FloatText(value=0.0, description='Annual Withdrawal (RM):', style={'description_width': 'initial'})

    output = widgets.Output()

    def on_value_change(change):
        with output:
            output.clear_output()
            calculated_balance = calculate_epf_balance(
                epfBalance_w.value,
                avgDividend_w.value,
                age_w.value,
                retirementAge_w.value,
                salary_w.value,
                bonus_w.value,
                increment_w.value,
                eeContribution_w.value,
                erContribution_w.value,
                monthlyWithdrawal_w.value,
                annualWithdrawal_w.value
            )
            print(f'EPF balance at retirement: RM{calculated_balance:,.2f}')

    # Link widgets to the function
    epfBalance_w.observe(on_value_change, names='value')
    avgDividend_w.observe(on_value_change, names='value')
    age_w.observe(on_value_change, names='value')
    retirementAge_w.observe(on_value_change, names='value')
    salary_w.observe(on_value_change, names='value')
    bonus_w.observe(on_value_change, names='value')
    increment_w.observe(on_value_change, names='value')
    eeContribution_w.observe(on_value_change, names='value')
    erContribution_w.observe(on_value_change, names='value')
    monthlyWithdrawal_w.observe(on_value_change, names='value')
    annualWithdrawal_w.observe(on_value_change, names='value')

    # Display initial calculation
    with output:
        initial_balance = calculate_epf_balance(
            epfBalance_w.value,
            avgDividend_w.value,
            age_w.value,
            retirementAge_w.value,
            salary_w.value,
            bonus_w.value,
            increment_w.value,
            eeContribution_w.value,
            erContribution_w.value,
            monthlyWithdrawal_w.value,
            annualWithdrawal_w.value
        )
        print(f'EPF balance at retirement: RM{initial_balance:,.2f}')

    # Arrange widgets in a VBox
    ui = widgets.VBox([
        epfBalance_w, avgDividend_w, age_w, retirementAge_w, salary_w, bonus_w,
        increment_w, eeContribution_w, erContribution_w, monthlyWithdrawal_w,
        annualWithdrawal_w, output
    ])
    display(ui)
```

```python
display_epf_calculator()
```
